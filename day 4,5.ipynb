```python
df['Date of birth']=pd.to_datetime(df['Date of birth'])
today=pd.Timestamp.today()
df['Age']=(
    (today-df['Date of birth']).dt.days//365
)
print(df[['First Name','Sex','Age']].head())
```

```python
pivot1=pd.pivot_table(
    df,
    values='Age',
    index='Sex'
)
print(pivot1)
```

```python
pivot2=pd.pivot_table(
    df,
    values='Age',
    index='Sex',
    aggfunc='count'
)
print(pivot2)
```

```python
pivot5=pd.pivot_table(
    df,
    values='Age',
    index='Sex',
    aggfunc=['count','max','min']
)
print(pivot5)
    
```

```python
pivot6=pd.pivot_table(
    df,
    values='Age',
    index='Sex',
    aggfunc='mean',

)
print(pivot6)
```

```python
pivot7=pd.pivot_table(
    df,
    values='User Id',
    index='Job Title',
    columns='Sex',
    aggfunc='count',
    fill_value=0
)
print(pivot7)
```

```python
df=pd.read_excel("datacleaning.xlsx")
df
```

```python
df.columns
```

```python
df.rename(columns={
    'Student ID':'stud_id',
    'Student Name ':'stud_nam',
    ' Department':'dept'
},inplace=True)
print(df.columns)
   
```

```python
print(df.dtypes)
```

```python
df['stud_id']=df['stud_id'].astype(str)
df['Age']=df['Age'].fillna(0)
df['Age']=df['Age'].astype(int)
print(df.dtypes)
```

```python
duplicates=df.duplicated()
print(duplicates)
```

```python
print(df.duplicated().sum())
```

```python
#dropping duplicate records
#befor
print(df.shape)
```

```python
#remove dauplicates
df.drop_duplicates(inplace=True)
```

```python
#after
print(df.shape)
```

```python
df.columns
```

```python
#string operations
df['Student Name ']=df['Student Name '].str.strip()
#convert to uppercase
df['Student Name ']=df['Student Name '].str.upper()
df['Student Name ']=df['Student Name '].str.title()
print(df['Student Name '].head())
```

```python
df[' Department']=df[' Department'].str.replace(
    'CSE',
    'Computer Science'
)
print(df[' Department'].head())
```

```python
df[' Department']=df[' Department'].replace({
    'CSE':'Computer Science',
    'ECE':'Electroni and communi eng',
    'EEE':'electrical and electronic',
    'IT':'information tech',
    'MECH':'Mechanical'
})
print(df[' Department'].head())
```

```python
#string length
df['Name_length']=df['Student Name '].str.len()
print(df[['Student Name ','Name_length']].head())
```

```python
df['First_Letter']=df['Student Name '].str[0]
print(df[['Student Name ','First_Letter']].head())
```
